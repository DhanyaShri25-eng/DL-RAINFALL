```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

# Replace with the path to your actual dataset
# For demonstration, we generate synthetic daily rainfall data
dates = pd.date_range(start='2020-01-01', end='2022-12-31', freq='D')
np.random.seed(42)
# Simulating rainfall data (mostly dry days with occasional rain peaks)
rainfall = np.random.exponential(scale=2.0, size=len(dates))
rainfall[rainfall < 1.0] = 0.0

df = pd.DataFrame({'Rainfall': rainfall}, index=dates)

# Creating lag features (e.g., past 5 days of rainfall to predict today's rainfall)
for i in range(1, 6):
    df[f'Lag_{i}'] = df['Rainfall'].shift(i)

# Drop rows with NaN values resulting from creating lag features
df = df.dropna()

# Define features (lags) and target
X = df[[f'Lag_{i}' for i in range(1, 6)]]
y = df['Rainfall']

# Perform train-test split (using a sequential split to respect time series nature)
split_idx = int(len(df) * 0.8)
X_train, X_test = X.iloc[:split_idx], X.iloc[split_idx:]
y_train, y_test = y.iloc[:split_idx], y.iloc[split_idx:]

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from sklearn.preprocessing import MinMaxScaler

# LSTMs are sensitive to the scale of the input features, so we normalize the data to [0, 1]
scaler_X = MinMaxScaler(feature_range=(0, 1))
scaler_y = MinMaxScaler(feature_range=(0, 1))

# Fit scalers on the training data and transform both sets
X_train_scaled = scaler_X.fit_transform(X_train)
X_test_scaled = scaler_X.transform(X_test)

# Reshape target arrays for the scaler
y_train_scaled = scaler_y.fit_transform(y_train.values.reshape(-1, 1))
y_test_scaled = scaler_y.transform(y_test.values.reshape(-1, 1))

# LSTM requires a 3D input of shape [samples, time steps, features]
# Since our features represent historical lags, we treat them as 1 time step with 5 features
X_train_lstm = np.reshape(X_train_scaled, (X_train_scaled.shape[0], 1, X_train_scaled.shape[1]))
X_test_lstm = np.reshape(X_test_scaled, (X_test_scaled.shape[0], 1, X_test_scaled.shape[1]))

print(f"LSTM Train Input Shape: {X_train_lstm.shape}")
print(f"LSTM Test Input Shape: {X_test_lstm.shape}")
```

```python
from tensorflow.keras.layers import Input

# Define the sequential model with an explicit Input layer
lstm_model = Sequential([
    Input(shape=(X_train_lstm.shape[1], X_train_lstm.shape[2])),
    LSTM(50, activation='relu', return_sequences=False),
    Dropout(0.2),
    Dense(1)  # Predicts continuous rainfall output
])

lstm_model.compile(optimizer='adam', loss='mean_squared_error')
lstm_model.summary()

# Train the network
history = lstm_model.fit(
    X_train_lstm, y_train_scaled,
    epochs=50,
    batch_size=8,
    validation_split=0.1,
    verbose=1
)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error

# Predict and reverse-transform back to standard millimeter units
y_pred_scaled = lstm_model.predict(X_test_lstm)
y_pred_lstm = scaler_y.inverse_transform(y_pred_scaled).flatten()

# Evaluate metrics
rmse_lstm = np.sqrt(mean_squared_error(y_test, y_pred_lstm))
mae_lstm = mean_absolute_error(y_test, y_pred_lstm)

print(f"LSTM Root Mean Squared Error (RMSE): {rmse_lstm:.2f} mm")
print(f"LSTM Mean Absolute Error (MAE): {mae_lstm:.2f} mm")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 6))
plt.plot(y_test.index, y_test.values, label='Actual Rainfall', color='blue', alpha=0.7)
plt.plot(y_test.index, y_pred_lstm, label='LSTM Predicted Rainfall', color='red', linestyle='-.', alpha=0.9)
plt.title('Daily Rainfall Prediction vs Actuals (Pune - LSTM Network)')
plt.xlabel('Date')
plt.ylabel('Rainfall (mm)')
plt.legend()
plt.grid(True)
plt.show()
```
